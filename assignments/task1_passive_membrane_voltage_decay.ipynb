```python
import numpy as np
import matplotlib.pyplot as plt
```

# 1. Calculate exponential function

```python
print(np.exp(0)) # e^0 = 1
print(np.exp(-1)) # e^-1
print(np.exp(-2)) # e^-2
```

```python
t = np.array([0, 12, 24])
print(np.exp(0.05776 * t))
```

# 2. Convert 20 kHz to time interval

```python
sampling_frequency = 20000 # samples per sec
dt_ms = 1000 / sampling_frequency # millisec, 측정 사이 시간간격

duration_ms = 250

t = np.arange(0, duration_ms, dt_ms) # 0-250미만까지 0.05ms 간격 시간생성

print("Time step:", dt_ms, "ms")
print("Number of samples:", len(t)) # 5001 (X) because arange()에서 끝값 250 X, 0-249.95까지만
print("First time:", t[0], "ms")
print("Last time:", t[-1], "ms")
```

# 3. Simulating & Plotting "Passive Membrane Voltage Decay"

```python
A = 60 # initial voltage in mV
half_life_ms = 12 # half-life in ms

gamma = np.log(2) / half_life_ms # decay constant

v = A * np.exp(-gamma * t)

print("Decay constant:", gamma, "per ms")
print("Initial voltage:", v[0], "mV")
print("Voltage at 12 ms:", v[240], "mv") # 240*0.05=12
print("Voltage at 24ms:", v[480], "mv") # 480*0.05=24
```

```python
plt.plot(t, v)
plt.xlabel("Time (ms)")
plt.ylabel("Voltage (mV)")
plt.title("Passive Membrane Voltage Decay")
plt.grid(True)
plt.show()
```
